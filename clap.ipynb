```python
from pathlib import Path
import time, re
import numpy as np, pyarrow as pa, pyarrow.parquet as pq
import torch

OUT_DIR = Path("/content/drive/MyDrive/tag-polysemy/clap-embeddings")
BASE = "clap_music_htsat"              # base name without extension
MAIN_F = OUT_DIR / f"{BASE}.parquet"     # your existing file
APP_F = OUT_DIR / f"{BASE}.append_{time.strftime('%Y%m%d-%H%M%S')}.parquet"
BATCH = 25

def parse_track_id(p):
    m = re.match(r"(\d+)", Path(p).name)
    return int(m.group(1)) if m else None

OUT_DIR.mkdir(parents=True, exist_ok=True)

# 1) Build a set of already-processed paths (fast: read only one column)
seen = set()
if MAIN_F.exists():
    t = pq.read_table(MAIN_F, columns=["path"])
    seen.update(t.column("path").to_pylist())
    print(f"Seen in main: {len(seen)}")

# Also check prior append files (optional but recommended)
for f in OUT_DIR.glob(f"{BASE}.append_*.parquet"):
    t = pq.read_table(f, columns=["path"])
    seen.update(t.column("path").to_pylist())
print(f"Total seen across all parts: {len(seen)}")

# 2) Filter the todo list
todo = [p for p in all_files if p not in seen]
print(f"Remaining to embed: {len(todo)}")

# 3) Write new rows to a *new* append file
if len(todo) == 0:
    print("Nothing to do.")
else:
    writer = None
    try:
        with torch.inference_mode():
            for i in range(0, len(todo), BATCH):
                batch_paths = todo[i:i+BATCH]
                embs = model.get_audio_embedding_from_filelist(x=batch_paths, use_tensor=False)  # (k,512)

                tab = pa.table({
                    "path": pa.array(batch_paths, type=pa.string()),
                    "track_id": pa.array([parse_track_id(p) for p in batch_paths], type=pa.int64()),
                    "embedding": pa.array(embs.tolist(), type=pa.list_(pa.float32(), 512)),
                })
                if writer is None:
                    writer = pq.ParquetWriter(APP_F, tab.schema, compression="zstd")
                writer.write_table(tab)

                if torch.cuda.is_available():
                    torch.cuda.empty_cache()
                if (i // BATCH) % 20 == 0:
                    print(f"  wrote {min(i+BATCH, len(todo))}/{len(todo)}")
    finally:
        if writer is not None:
            writer.close()
    print(f"Appended {len(todo)} rows to {APP_F}")
```
